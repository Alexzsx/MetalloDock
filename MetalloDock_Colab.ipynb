```
%%bash
set -euo pipefail

/content/metallodock_colab/env/bin/python - <<'PY'
import importlib.metadata as metadata
import torch
import torch_geometric
import torch_spline_conv
from torch_scatter import scatter_add
from torch_sparse import SparseTensor
from torch_cluster import knn_graph
from torch_geometric.nn import GraphNorm

assert torch.__version__ == "2.4.0+cu121"
assert torch.cuda.is_available()

print("PyTorch：", torch.__version__)
print("PyG：", torch_geometric.__version__)

for package in [
    "torch-scatter", "torch-sparse",
    "torch-cluster", "torch-spline-conv"
]:
    print(package, metadata.version(package))

# 1. GPU scatter
src = torch.tensor([1., 2., 3., 4.], device="cuda")
idx = torch.tensor([0, 0, 1, 1], device="cuda")
out = scatter_add(src, idx, dim=0)
assert torch.allclose(
    out, torch.tensor([3., 7.], device="cuda")
)
print("GPU scatter：通过")

# 2. GPU稀疏矩阵乘法
row = torch.tensor([0, 1], device="cuda")
col = torch.tensor([0, 1], device="cuda")
matrix = SparseTensor(
    row=row, col=col,
    value=torch.ones(2, device="cuda"),
    sparse_sizes=(2, 2),
)
dense = torch.randn(2, 3, device="cuda")
assert torch.allclose(matrix.matmul(dense), dense)
print("GPU sparse：通过")

# 3. GPU近邻图
points = torch.randn(6, 3, device="cuda")
edges = knn_graph(points, k=2, loop=False)
assert edges.shape == (2, 12)
print("GPU knn_graph：通过")

# 4. 模型会用到的GraphNorm
norm = GraphNorm(4).cuda()
result = norm(torch.randn(6, 4, device="cuda"))
assert torch.isfinite(result).all().item()

torch.cuda.synchronize()
print("GPU GraphNorm：通过")
print("PyG环境验证：全部通过")
PY
```

```
from pathlib import Path
from datetime import datetime, timezone
import subprocess
import shutil
import hashlib
import json

task = Path("/content/metallodock_colab")
py = task / "env/bin/python"
repo = task / "MetalloDock"
run = Path((task / "logs/latest_pa_run.txt").read_text().strip())

stamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%SZ")
dest = task / f"pa_checkpoint_{stamp}"
dest.mkdir()

def capture(name, command):
    result = subprocess.run(command, capture_output=True, text=True)
    (dest / name).write_text(
        result.stdout + "\n" + result.stderr,
        encoding="utf-8"
    )
    if result.returncode:
        raise RuntimeError(f"{name} 导出失败，请检查输出")

capture("pip_freeze.txt", [str(py), "-m", "pip", "freeze"])
capture("conda_explicit.txt", [
    str(task / "bin/micromamba"),
    "--root-prefix", str(task / "mamba_root"),
    "list", "--prefix", str(task / "env"), "--explicit"
])
capture("source_commit.txt", [
    "git", "-C", str(repo), "rev-parse", "HEAD"
])
capture("source_diff.patch", [
    "git", "-C", str(repo), "diff", "HEAD"
])
capture("gpu.txt", ["nvidia-smi"])

shutil.copytree(task / "logs", dest / "logs")
shutil.copytree(run, dest / "official_pa_run")

for path in task.glob("*specs.txt"):
    shutil.copy2(path, dest / path.name)
for path in task.glob("*constraints*.txt"):
    shutil.copy2(path, dest / path.name)

manifest = {
    str(p.relative_to(dest)): hashlib.sha256(p.read_bytes()).hexdigest()
    for p in sorted(dest.rglob("*")) if p.is_file()
}
(dest / "SHA256_MANIFEST.json").write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

archive = Path(shutil.make_archive(str(dest), "zip", root_dir=dest))
digest = hashlib.sha256(archive.read_bytes()).hexdigest()
print("备份：", archive.name)
print("大小：", round(archive.stat().st_size / 1024**2, 2), "MB")
print("SHA-256：", digest)

from google.colab import files
files.download(str(archive))
```

```
from google.colab import files
from pathlib import Path
from datetime import datetime, timezone
import hashlib
import io
import os
import shutil
import subprocess
import zipfile

task = Path("/content/metallodock_colab")
bundle = task / "metallodock_colab_diagnostics_v1"

uploaded = files.upload()
assert len(uploaded) == 1, "请只上传三例输入 ZIP"
data = next(iter(uploaded.values()))

expected = "3194c9c621151638e5a50dd4b8a4b4caf62c51a08104124fd2ea6e5f3fa46dde"
assert hashlib.sha256(data).hexdigest() == expected, "上传文件哈希不匹配"

with zipfile.ZipFile(io.BytesIO(data)) as z:
    assert z.testzip() is None
    z.extractall(task)

cmd = [
    str(task / "env/bin/python"), "-u",
    str(bundle / "run_diagnostics.py"),
    "--repo", str(task / "MetalloDock")
]

env = {
    **os.environ,
    "OMP_NUM_THREADS": "1",
    "MKL_NUM_THREADS": "1",
    "PYTHONUNBUFFERED": "1",
    "PYTHONDONTWRITEBYTECODE": "1"
}

with (bundle / "COLAB_SESSION.log").open("a") as log:
    proc = subprocess.Popen(
        cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
        text=True, bufsize=1, env=env
    )
    for line in proc.stdout:
        print(line, end="", flush=True)
        log.write(line)
        log.flush()
    code = proc.wait()

print("\n运行退出码：", code)

stamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%SZ")
archive = Path(shutil.make_archive(
    str(task / f"metallodock_threecase_results_{stamp}"),
    "zip", root_dir=bundle
))
print("结果包：", archive.name)
files.download(str(archive))
```
